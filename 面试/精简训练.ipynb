## 1. 反转链表 (206) — Easy ⭐⭐⭐⭐⭐
链表题祖宗，不会写直接挂。**考点**：三指针迭代

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverseList(head):
    prev = None
    cur = head
    while cur:
        nxt = cur.next   # 1. 保存下一个
        cur.next = prev  # 2. 反转指针
        prev = cur       # 3. prev前进
        cur = nxt        # 4. cur前进
    return prev

def to_list(head):
    res = []
    while head: res.append(head.val); head = head.next
    return res

head = ListNode(1, ListNode(2, ListNode(3)))
print(to_list(reverseList(head)))  # [3,2,1]
```

## 2. LRU缓存 (146) — Medium ⭐⭐⭐⭐⭐
设计题最高频。**考点**：哈希表 + 双向链表，get/put O(1)

```python
class DLinkedNode:
    def __init__(self, key=0, val=0):
        self.key = key; self.val = val
        self.prev = None; self.next = None

class LRUCache:
    def __init__(self, capacity):
        self.cap = capacity
        self.cache = {}
        self.head = DLinkedNode()
        self.tail = DLinkedNode()
        self.head.next = self.tail
        self.tail.prev = self.head
    
    def _remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev
    
    def _add_to_head(self, node):
        node.next = self.head.next
        node.prev = self.head
        self.head.next.prev = node
        self.head.next = node
    
    def get(self, key):
        if key not in self.cache: return -1
        node = self.cache[key]
        self._remove(node)
        self._add_to_head(node)
        return node.val
    
    def put(self, key, val):
        if key in self.cache:
            self._remove(self.cache[key])
        node = DLinkedNode(key, val)
        self.cache[key] = node
        self._add_to_head(node)
        if len(self.cache) > self.cap:
            last = self.tail.prev
            self._remove(last)
            del self.cache[last.key]

lru = LRUCache(2)
lru.put(1,1); lru.put(2,2)
print(lru.get(1))  # 1
lru.put(3,3)
print(lru.get(2))  # -1 (被淘汰)
```

## 3. 两数之和 (1) — Easy
哈希表入门。**考点**：一遍遍历 + 查字典

```python
def twoSum(nums, target):
    seen = {}
    for i, v in enumerate(nums):
        need = target - v
        if need in seen:
            return [seen[need], i]
        seen[v] = i

print(twoSum([2,7,11,15], 9))  # [0,1]
```

## 4. 三数之和 (15) — Medium
双指针经典。**考点**：排序 + 固定一个 + 双指针两头走 + 去重

```python
def threeSum(nums):
    nums.sort()
    n = len(nums)
    res = []
    for i in range(n - 2):
        if nums[i] > 0: break
        if i > 0 and nums[i] == nums[i-1]: continue
        l, r = i + 1, n - 1
        while l < r:
            s = nums[i] + nums[l] + nums[r]
            if s < 0: l += 1
            elif s > 0: r -= 1
            else:
                res.append([nums[i], nums[l], nums[r]])
                while l < r and nums[l] == nums[l+1]: l += 1
                while l < r and nums[r] == nums[r-1]: r -= 1
                l += 1; r -= 1
    return res

print(threeSum([-1,0,1,2,-1,-4]))  # [[-1,-1,2],[-1,0,1]]
```

## 5. 无重复字符的最长子串 (3) — Medium
滑动窗口。**考点**：双指针 + 哈希集，右指针扩张，左指针收缩

```python
def lengthOfLongestSubstring(s):
    seen = set()
    l = 0; ans = 0
    for r in range(len(s)):
        while s[r] in seen:
            seen.remove(s[l])
            l += 1
        seen.add(s[r])
        ans = max(ans, r - l + 1)
    return ans

print(lengthOfLongestSubstring("abcabcbb"))  # 3
```

## 6. 有效的括号 (20) — Easy
栈。**考点**：左括号入栈，右括号匹配栈顶

```python
def isValid(s):
    pair = {')': '(', ']': '[', '}': '{'}
    stack = []
    for ch in s:
        if ch not in pair:
            stack.append(ch)
        else:
            if not stack or stack.pop() != pair[ch]:
                return False
    return not stack

print(isValid("()[]{}"))  # True
print(isValid("([)]"))    # False
```

## 7. 层序遍历 (102) — Medium
BFS模板。**考点**：队列 + 按层记录size

```python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def levelOrder(root):
    if not root: return []
    res, q = [], deque([root])
    while q:
        level = []
        for _ in range(len(q)):  # 关键：先记录当前层大小
            node = q.popleft()
            level.append(node.val)
            if node.left: q.append(node.left)
            if node.right: q.append(node.right)
        res.append(level)
    return res
```

## 8. 最近公共祖先 (236) — Medium ⭐⭐⭐⭐⭐
后序遍历。**考点**：自底向上，第一个左右各找到一个的节点=LCA

```python
def lowestCommonAncestor(root, p, q):
    if not root or root == p or root == q:
        return root
    left = lowestCommonAncestor(root.left, p, q)
    right = lowestCommonAncestor(root.right, p, q)
    if left and right:
        return root  # p和q在两侧
    return left or right  # 都在一侧
```

## 9. 最大子数组和 (53) — Medium
DP入门。**考点**：Kadane算法 dp[i]=max(nums[i], dp[i-1]+nums[i])

```python
def maxSubArray(nums):
    dp = nums[0]
    ans = nums[0]
    for i in range(1, len(nums)):
        dp = max(nums[i], dp + nums[i])  # 接or重新开始
        ans = max(ans, dp)
    return ans

print(maxSubArray([-2,1,-3,4,-1,2,1,-5,4]))  # 6
```

## 10. 爬楼梯 (70) — Easy
斐波那契DP。**考点**：dp[i]=dp[i-1]+dp[i-2]，空间优化到O(1)

```python
def climbStairs(n):
    if n <= 2: return n
    a, b = 1, 2
    for _ in range(3, n+1):
        a, b = b, a + b
    return b

print(climbStairs(5))  # 8
```

## 11. 买卖股票最佳时机 (121) — Easy
贪心。**考点**：遍历时记录最低价，每天算利润

```python
def maxProfit(prices):
    min_price = float('inf')
    max_profit = 0
    for p in prices:
        min_price = min(min_price, p)
        max_profit = max(max_profit, p - min_price)
    return max_profit

print(maxProfit([7,1,5,3,6,4]))  # 5
```

## 12. 搜索旋转排序数组 (33) — Medium
二分变体。**考点**：判断mid落在左半还是右半，再判断target在哪边

```python
def search(nums, target):
    l, r = 0, len(nums) - 1
    while l <= r:
        mid = l + (r - l) // 2
        if nums[mid] == target: return mid
        if nums[l] <= nums[mid]:  # 左半有序
            if nums[l] <= target < nums[mid]:
                r = mid - 1
            else:
                l = mid + 1
        else:  # 右半有序
            if nums[mid] < target <= nums[r]:
                l = mid + 1
            else:
                r = mid - 1
    return -1

print(search([4,5,6,7,0,1,2], 0))  # 4
```

## 13. 数组第K大元素 (215) — Medium
快排partition / 堆。**考点**：quickselect O(n)平均

```python
import random

def findKthLargest(nums, k):
    def partition(l, r, pivot_idx):
        pivot = nums[pivot_idx]
        nums[pivot_idx], nums[r] = nums[r], nums[pivot_idx]
        store = l
        for i in range(l, r):
            if nums[i] > pivot:  # 降序，大的放左边
                nums[store], nums[i] = nums[i], nums[store]
                store += 1
        nums[r], nums[store] = nums[store], nums[r]
        return store
    
    def quickselect(l, r, k):
        if l == r: return nums[l]
        pivot_idx = random.randint(l, r)
        pos = partition(l, r, pivot_idx)
        if pos == k: return nums[pos]
        elif pos < k: return quickselect(pos+1, r, k)
        else: return quickselect(l, pos-1, k)
    
    return quickselect(0, len(nums)-1, k-1)

print(findKthLargest([3,2,1,5,6,4], 2))  # 5
```

## 14. 二分查找 (704) — Easy
二分模板。**考点**：while l<=r, mid=l+(r-l)//2 防溢出

```python
def search(nums, target):
    l, r = 0, len(nums) - 1
    while l <= r:
        mid = l + (r - l) // 2
        if nums[mid] == target: return mid
        elif nums[mid] < target: l = mid + 1
        else: r = mid - 1
    return -1

print(search([-1,0,3,5,9,12], 9))  # 4
```

## 15. 快速排序 ⭐⭐⭐⭐⭐
面试直接让你写的排序。**考点**：随机pivot + partition + 递归

```python
import random

def quicksort(nums):
    def partition(l, r):
        pivot_idx = random.randint(l, r)
        pivot = nums[pivot_idx]
        nums[pivot_idx], nums[r] = nums[r], nums[pivot_idx]
        i = l  # i左边都是<pivot的
        for j in range(l, r):
            if nums[j] < pivot:
                nums[i], nums[j] = nums[j], nums[i]
                i += 1
        nums[i], nums[r] = nums[r], nums[i]
        return i
    
    def _sort(l, r):
        if l >= r: return
        p = partition(l, r)
        _sort(l, p-1)
        _sort(p+1, r)
    
    _sort(0, len(nums)-1)
    return nums

print(quicksort([3,1,4,1,5,9,2,6]))  # [1,1,2,3,4,5,6,9]
```

## 16. 逻辑回归交叉熵推导 🔴🔴🔴

**一句话**：逻辑回归加权求和后过sigmoid输出概率，用交叉熵衡量预测和真实的差距。

**推导链**：
1. 线性：z = w^T·x + b
2. Sigmoid：ŷ = 1/(1+e^{-z})，输出∈(0,1)
3. 似然：P(y|x) = ŷ^y·(1-ŷ)^{1-y}
4. 取负对数：J(w) = -(1/m)·Σ[y_i·log(ŷ_i) + (1-y_i)·log(1-ŷ_i)]
5. **关键梯度（必须背）**：∂J/∂w = (1/m)·Σ(ŷ_i - y_i)·x_i

**为什么交叉熵不用MSE？** MSE+sigmoid→梯度消失（ŷ→0或1时梯度→0）+非凸优化

**项目关联**：GNN项目LightGBM二分类目标函数底层就是逻辑回归交叉熵。区别是LGB用boosting tree做特征变换，最后一层sigmoid出概率。

## 17. L1 vs L2 正则化 🔴

**一句话**：L1让不重要的权重直接变0（特征选择），L2让所有权重变小但不为0（防过拟合）。

**公式**：L1: J+λΣ|w_i|  L2: J+(λ/2)Σw_i²

**几何**：L1约束=菱形（最优解在顶点→稀疏）L2约束=圆（均匀缩小）

**梯度**：L1梯度=sign(w)固定推往0；L2梯度=2w，w小时推不动

**贝叶斯**：L2=高斯先验 L1=拉普拉斯先验

**项目**：LightGBM的reg_alpha(L1)+reg_lambda(L2)，实际调参两者一起用=ElasticNet

## 18. GBDT为什么拟合负梯度？🔴🔴🔴

**一句话**：GBDT用CART回归树拟合损失函数的负梯度方向——这是在函数空间做梯度下降。

**推导**：想找f_t使L(y,F_{t-1}+f_t)最小 → 泰勒展开L≈L(F_{t-1})+∂L/∂F·f_t → 让第二项最小→f_t=-∂L/∂F（负梯度）

**MSE损失**：负梯度=y-F=残差（最直观）
**交叉熵**：负梯度=y-sigmoid(F)=残差（形式相同）

**项目**：LightGBM每棵树都在拟合前面所有树的残差之和，最终预测是所有树加权和

## 19. XGBoost vs LightGBM 🔴🔴

**XGBoost 5大改进（相比GBDT）**：①二阶泰勒展开 ②正则化(叶子数+权重) ③列采样 ④缺失值自动处理 ⑤近似算法

**LightGBM 4大改进（相比XGBoost）**：
1. **直方图算法**：连续值→K个bin(默认255)，不排序
2. **Leaf-wise生长**：每次分裂增益最大的叶子(vs Level-wise逐层)，更深更准
3. **GOSS**：保留大梯度样本+随机采小梯度
4. **EFB**：互斥特征捆绑降维

**项目**：GNN项目选LightGBM因为比XGBoost快一个量级，200+维特征，直方图+leaf-wise省大量时间

## 20. BatchNorm 训练 vs 推理 🔴

**训练时**：对当前mini-batch算均值μ_B和方差σ²_B→归一化→缩放平移(y=γx̂+β，γβ可学)

**推理时**：用训练过程中全局移动平均的μ和σ²（不是当前batch的！）

**作用**：加速收敛（loss曲面更平滑）、缓解梯度消失、允许更大学习率、轻微正则化

**LayerNorm vs BatchNorm**：BN对batch维度归一化→CNN/MLP用；LN对feature维度归一化→RNN/Transformer用（序列长度不一致+batch统计不稳定）

## 21. Transformer Self-Attention 🔴🔴🔴

**公式**：Attention(Q,K,V) = softmax(QK^T/√d_k)·V

**Q/K/V**：输入X分别乘W^Q/W^K/W^V得到。类比数据库：Query查Key→得权重→加权取Value

**为什么要除以√d_k？** QK^T结果方差=d_k，d_k大→点积大→softmax趋近one-hot(梯度0)→除以√d_k让方差=1

**Multi-Head**：多组Q/K/V投影到不同子空间→不同头关注不同方面（语法/语义/位置）

**Position Encoding**：Transformer没有RNN的时序→注入位置信息(sin/cos或可学习)

## 22. 反向传播链式法则 🔴

**一句话**：梯度从输出层逐层往前传，每层乘权重的转置和激活函数的导数。

**2层MLP推导**（X→ReLU(hidden)→Sigmoid(output)→交叉熵）：
1. ∂L/∂W₂ = (ŷ-y)·h^T  ← 输出层梯度=误差×隐层输出
2. ∂L/∂h = W₂^T·(ŷ-y)
3. ∂L/∂W₁ = [W₂^T·(ŷ-y) ⊙ 1{z₁>0}]·X^T  ← ReLU导数=indicator

**核心记忆**：链式法则本质=复合函数求导，反向传播=从后往前逐层计算

## 23. AUC物理含义

**一句话**：随机选一个正样本和一个负样本，正样本预测值>负样本预测值的概率。

AUC=1→所有正样本排在负样本前面（完美）
AUC=0.5→随机排序

**AUC vs F1**：AUC衡量全局排序质量（不依赖阈值），F1依赖阈值选择

**项目**：GNN场景2用AUC评估全局排序；场景1用MRR（关心第一个正确答案排第几）

## 24. Wide&Deep / DeepFM 🔴

**Wide&Deep**：Wide侧=原始特征+人工交叉→**记忆**(记住历史共现)；Deep侧=Embedding→MLP→**泛化**(学到没见过的组合)

**DeepFM改进**：用FM层自动学习二阶特征交叉替换Wide侧的人工交叉→不需要人工特征工程；FM和Deep共享Embedding

## 25. DIN 注意力机制 🔴

## 27. 过拟合/欠拟合判断 🔴

**过拟合**：train loss↓但val loss↑（差距拉大）→模型在背训练集→减特征/加正则/早停/加数据

**欠拟合**：train和val loss都高且接近→模型太简单→加特征/更复杂模型/减正则

**刚好**：train和val都收敛且差距小

**项目**：GNN训练时监控train loss和val MRR，val MRR开始下降的那个epoch就是早停点

## 28. ReLU vs Sigmoid

**ReLU三优势**：①导数大(x>0时导数=1，不衰减) ②计算简单(max操作) ③稀疏激活(x<0输出0→类似dropout)

**Sigmoid致命伤**：导数最大0.25→深层网络每层乘以0.25→指数级梯度消失

**Dead ReLU**：权重更新后对所有输入都输出负→神经元永久"死亡"→用LeakyReLU/降低学习率/He初始化

## 29. LSTM三个门 🔴🔴

| 门 | 作用 |
|----|------|
| 遗忘门 f_t | 决定丢弃哪些旧信息（0=全忘） |
| 输入门 i_t | 决定哪些新信息写入cell state |
| 输出门 o_t | 决定输出多少cell state |

**Cell State更新**：C_t = f_t⊙C_{t-1} + i_t⊙C̃_t → 加法形式！→ 梯度通过C_t直达C_{t-1}（不需要乘W_h）→ 缓解梯度消失（Constant Error Carousel）

**GRU**：合并遗忘门+输入门=更新门z_t，2个门参数更少→训练更快

## 30. Adam优化器

**本质**：Momentum + RMSprop 的结合
- m_t = β₁·m_{t-1}+(1-β₁)·g_t（一阶动量）
- v_t = β₂·v_{t-1}+(1-β₂)·g_t²（二阶自适应学习率）
- 偏差修正 m̂_t=m_t/(1-β₁^t) → θ_t=θ_{t-1}-α·m̂_t/(√v̂_t+ε)

**默认参数**：α=0.001, β₁=0.9, β₂=0.999, ε=1e-8

**Adam vs SGD**：Adam收敛快、调参少；SGD+Momentum调参好泛化更好（ImageNet大任务常用）

**实践**：先用Adam快速迭代，后用SGD精调
